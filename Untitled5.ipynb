```python
###与上一单元完全等价
def classify_gaps(gap_list):
    result = []  # 1. 先拿一个空坩埚 []
    for gap in gap_list:  # 2. 遍历原料
        if gap > 3.0:
            result.append("绝缘体")  # 3. 把成品扔进坩埚
        else:
            result.append("半导体")
    return result  # 4. 递出坩埚
print(result)
```

```python
class UnitCell:
    def __init__(self,a,b,c):
        self.a=a
        self.b=b
        self.c=c
    def volume(self):
        V=self.a*self.b*self.c
        return V
    def show_info(self):
        print(f"晶胞参数a={self.a} b={self.b} c={self.c}")
si_cell = UnitCell(5.43, 5.43, 5.43)
si_cell.show_info()
vol = si_cell.volume()
print(f"晶胞体积是 {vol:.3f}")
```

```python
class BandStructure:
    def __init__(self,material_name,band_gap,is_direct):
        self.material_name=material_name
        self.band_gap=band_gap
        self.is_direct=is_direct
    def show_info(self):
        print(f"材料{self.material_name},带隙{self.band_gap},直接带隙{self.is_direct}")
    def classify(self):
        if self.band_gap<=0:
            return "金属"
        elif self.band_gap<=3.0:
            return "半导体"
        else:
            return "绝缘体"
mos2=BandStructure("MoS2",1.8,True)
mos2.show_info()
print(mos2.classify())
```

```python
class BandStructure:
    def __init__(self,material_name,band_gap,is_direct):
        self.material_name=material_name
        self.band_gap=band_gap
        self.is_direct=is_direct
    def show_info(self):
        print(f"材料{self.material_name},带隙{self.band_gap},直接带隙{self.is_direct}")
    def classify(self):
        if self.band_gap<=0:
            return "金属"
        elif self.band_gap<=3.0:
            return "半导体"
        else:
            return "绝缘体"
mos2=BandStructure("MoS2",1.8,True)
gan=BandStructure("GaN",3.4,True)
gaas=BandStructure("GaAs",1.42,True)
materials_db=[mos2,gan,gaas]
wide_band_gaps=[mat.material_name for mat in materials_db if mat.band_gap >2.0]
print(wide_band_gaps)
```

```python
class Material:
    def __init__(self, name, lattice_a):
        self.name=name
        self.lattice_a=lattice_a
class TwoDMaterial(Material):
    def __init__(self, name, lattice_a, layer_count):
        super().__init__(name, lattice_a)
        self.layer_count=layer_count
mos2_2d = TwoDMaterial("MoS2", 3.16, 1)
print(f"材料: {mos2_2d.name}, 晶格常数: {mos2_2d.lattice_a}, 层数: {mos2_2d.layer_count}")
```
